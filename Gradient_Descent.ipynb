```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
students = pd.read_csv('students_dataset.csv')
```

```python
students
```

```python
X = students.loc[:, ['CGPA', 'IQ']]
y = students.loc[:, ['Package_LPA']]
```

```python
#Splitting dataset into X_train, X_test, y_train and y_test
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.3, random_state=42)
```

```python
#Batch Gradient Descent Regressor
class BatchGradientDescent:
  def __init__(self):
    self.coef_ = None
    self.intercept_ = None
  def fit(self, X_train, y_train):
    n = X_train.shape[0]
    no_predictors = X_train.shape[1]
    X_train = np.insert(X_train, 0, 1, axis =1)
    y_train = np.array(y_train)
    n_iterations = 100000
    eta = 0.001
    theta = np.random.randn(no_predictors+1).reshape((no_predictors+1,1))
    for i in range(n_iterations):
      gradient_vector = (2/n)*(X_train.T.dot(X_train.dot(theta) -y_train))
      theta = theta - (eta*gradient_vector)
    self.coef_ = theta[1:].reshape((y_train.shape[1], no_predictors))
    self.intercept_ = theta[0]
```

```python
#Scaling X_train dataset before fitting
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_train)

#Finding best fit parameters using sklearn model
from sklearn.linear_model import LinearRegression
sk_regressor = LinearRegression()
sk_regressor.fit(X_scaled, y_train)

#Finding best fit parameters using custom model
bgd_regressor = BatchGradientDescent()
bgd_regressor.fit(X_scaled, y_train)

#Comparing closed form solution and gradient descent
print('Estimated parameters from sklearn regressor')
print('weights:', sk_regressor.coef_)
print('intercept:', sk_regressor.intercept_)
print('Estimated parameters from custom regressor')
print('weights:', bgd_regressor.coef_)
print('intercept:', bgd_regressor.intercept_)
```

```python
#Stochashtic Gradient Descent
class StochasticGradientDescent:
  def __init__(self):
    self.coef_ = None
    self.intercept_ = None
  def fit(self, X_train, y_train):
    n = X_train.shape[0]
    no_predictors = X_train.shape[1]
    X_train = np.insert(X_train, 0, 1, axis =1)
    y_train = np.array(y_train)
    theta = np.random.randn(no_predictors+1).reshape((no_predictors+1, 1))

    #Learning rate function
    def learning_rate(t):
      t0, t1 = 5,50
      return t0 / (t+t1)

    n_epochs = 1000    #No of rounds
    for epoch in range(n_epochs):
      for i in range(n):
        rand_idx = np.random.randint(n)
        xi = np.array(X_train)[rand_idx].reshape((1,-1))   #Row vector of 1 instance
        yi = np.array(y_train)[rand_idx]
        gradient_vector = 2*(xi.T.dot(xi.dot(theta) - yi))
        eta = learning_rate(epoch*n + i)
        theta = theta - (eta*gradient_vector)
    self.coef_ = theta[1:].reshape(1,-1)
    self.intercept_ = theta[0]
```

```python
#Scaling X_train dataset before fitting
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_train)

#Finding best fit parameters using sklearn model
from sklearn.linear_model import LinearRegression
sk_regressor = LinearRegression()
sk_regressor.fit(X_scaled, y_train)

#Finding best fit parameters using custom model
sgd_regressor = StochasticGradientDescent()
sgd_regressor.fit(X_scaled, y_train)

#Comparing closed form solution and gradient descent
print('Estimated parameters from sklearn regressor')
print('weights:', sk_regressor.coef_)
print('intercept:', sk_regressor.intercept_)
print('Estimated parameters from custom regressor')
print('weights:', sgd_regressor.coef_)
print('intercept:', sgd_regressor.intercept_)
```

```python
#Mini Batch Gradient Descent
#Stochashtic Gradient Descent
class MiniBatchGradientDescent:
  def __init__(self):
    self.coef_ = None
    self.intercept_ = None
  def fit(self, X_train, y_train):
    n = X_train.shape[0]
    no_predictors = X_train.shape[1]
    X_train = np.insert(X_train, 0, 1, axis =1)
    y_train = np.array(y_train)
    theta = np.random.randn(no_predictors+1).reshape((no_predictors+1, 1))

    #Learning rate function
    def learning_rate(t):
      t0, t1 = 5,50
      return t0 / (t+t1)

    n_epochs = 1000    #No of rounds
    for epoch in range(n_epochs):
      for i in range(n):
        rand_idx = np.random.randint(n,size=5)
        xi = np.array(X_train)[rand_idx]  #Row vector of 1 instance
        yi = np.array(y_train)[rand_idx]
        gradient_vector = 2*(xi.T.dot(xi.dot(theta) - yi))
        eta = learning_rate(epoch*n + i)
        theta = theta - (eta*gradient_vector)
    self.coef_ = theta[1:].reshape(1,-1)
    self.intercept_ = theta[0]
```

```python
#Scaling X_train dataset before fitting
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_train)

#Finding best fit parameters using sklearn model
from sklearn.linear_model import LinearRegression
sk_regressor = LinearRegression()
sk_regressor.fit(X_scaled, y_train)

#Finding best fit parameters using custom model
mgd_regressor = MiniBatchGradientDescent()
mgd_regressor.fit(X_scaled, y_train)

#Comparing closed form solution and gradient descent
print('Estimated parameters from sklearn regressor')
print('weights:', sk_regressor.coef_)
print('intercept:', sk_regressor.intercept_)
print('Estimated parameters from custom regressor')
print('weights:', mgd_regressor.coef_)
print('intercept:', mgd_regressor.intercept_)
```
